```python
# Pull x variables from data frame
x_variables = fb_full_df[["actual_dist", "hang_time", "descent_angle_deg", "attack_angle_deg"]].to_numpy()

# Use the model on those variables
fb_actual_predict = logistic_regression_model.predict_proba(x_variables)

# Take Catch Prob % from projections and add it to the data frame
fb_actual_predict_df = pl.DataFrame({ # https://stackoverflow.com/questions/79622730/split-an-array-in-a-polars-dataframe-into-regular-columns
    "log_prob": fb_actual_predict
             })
# Isolate the % of a catch
catch_prob = fb_actual_predict_df.select(
                                    pl.col("log_prob").arr.get(1).alias("catch_prob")
                                    )
# Add it to the df
fb_actual_prob_df = pl.concat([fb_full_df, catch_prob],
                                    how = "horizontal")



# Do the same thing but using the avg starting position instead of the actual position
x_avg_variables = fb_full_df[["avg_dist_split", "hang_time","descent_angle_deg", "attack_angle_avg_deg"]].to_numpy()

# Use the model on the avg variables
fb_avg_predict = logistic_regression_model.predict_proba(x_avg_variables)

fb_avg_predict_df = pl.DataFrame({ #https://stackoverflow.com/questions/79622730/split-an-array-in-a-polars-dataframe-into-regular-columns
    "log_prob": fb_avg_predict
             })
# Isolate the % of a catch
avg_prob = fb_avg_predict_df.select(
                                    pl.col("log_prob").arr.get(1).alias("avg_prob")
                                    )
# Add it to the df
fb_avg_prob_df = pl.concat([fb_actual_prob_df, avg_prob],
                                    how = "horizontal")
# Select relevant columns and add player ID
fb_delta_lf_df = fb_avg_prob_df\
                            .select(
                                ["game_string", "play_per_game", "actual_dist", "avg_dist_split", "hang_time", "attack_angle_deg", "attack_angle_avg_deg", "descent_angle_deg", "catch_status", "catch_prob", "avg_prob"
                                ])\
                            .with_columns(
                                prob_delta = (pl.col("catch_prob") - pl.col("avg_prob")),
                                player_id = of_pos
                            ).sort(
                    pl.col("prob_delta").abs(),
                    descending = True
                )
```
